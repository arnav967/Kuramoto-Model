```python
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML


plt.rcParams['animation.embed_limit'] = 60.0


N = 80
G = nx.cycle_graph(N)


num_shortcuts = 10 
added = 0
while added < num_shortcuts:
     
    u, v = np.random.choice(range(N), 2, replace=False)
    

    if not G.has_edge(u, v):
        G.add_edge(u, v)
        added += 1

pos = nx.circular_layout(G)

K = 10.0    
Dt = 0.05    
steps = 3000 

def initialize_ring_80():
    global theta, omega
    np.random.seed(42)
    theta = np.random.uniform(0, 2*np.pi, N)
    # A bit of frequency spread to keep it realistic
    omega = 2.0 + np.random.uniform(-1.0, 1.0, N)


initialize_ring_80()


fig, ax = plt.subplots(figsize=(9, 9))
ax.axis('off')


nx.draw_networkx_edges(G, pos, ax=ax, edge_color='#dddddd', alpha=0.3)
node_collection = nx.draw_networkx_nodes(G, pos, ax=ax, node_size=150)

def update(frame):
    global theta
    d_theta = np.zeros(N)
    
    
    for i in G.nodes():
        neighbors = list(G.neighbors(i))
      
        coupling = sum(np.sin(theta[j] - theta[i]) for j in neighbors) / 2.0
        d_theta[i] = omega[i] + (K * coupling)
    
    theta = theta + d_theta * Dt
    
   
    new_colors = [plt.cm.hsv((t % (2*np.pi)) / (2*np.pi)) for t in theta]
    node_collection.set_facecolor(new_colors)
    
    ax.set_title(f"Ring System (N=80) | Step: {frame}", fontsize=15)
    return node_collection,


ani = FuncAnimation(fig, update, frames=range(0, steps, 10), interval=50, blit=True)


plt.close()
HTML(ani.to_html5_video())
```
