```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint
from matplotlib.animation import FuncAnimation
from IPython.display import HTML


N = 2
K = 3.0     ### change k to 1 or 3 to see either simulation        
omega = np.array([0.5, 2.5]) 
t = np.linspace(0, 20, 400)
theta0 = np.array([0.0, np.pi])


def kuramoto(theta, t, omega, K, N):
    d_theta21 = omega[1] - omega[0] - (2 * K / N) * np.sin(theta[1] - theta[0]) #
    
    d_theta = np.zeros(N)
    for i in range(N):
        coupling = (K / N) * np.sum(np.sin(theta - theta[i]))
        d_theta[i] = omega[i] + coupling
    return d_theta


sol = odeint(kuramoto, theta0, t, args=(omega, K, N))
sol_wrapped = sol % (2 * np.pi)

#Visualization
fig, ax = plt.subplots(figsize=(6, 4))
ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-1.5, 1.5)
ax.set_aspect('equal')
ax.axis('off')

pos = np.array([[-0.8, 0], [0.8, 0]])

ax.plot(pos[:, 0], pos[:, 1], color='gray', lw=2, zorder=1)

circles = [plt.Circle((pos[i, 0], pos[i, 1]), 0.3, ec='black', zorder=2) for i in range(N)]
for circle in circles:
    ax.add_patch(circle)

def update(frame):
    phases = sol_wrapped[frame]
    for i, circle in enumerate(circles):
        # Map phase to HSV color wheel
        color = plt.cm.hsv(phases[i] / (2 * np.pi))
        circle.set_facecolor(color)
    return circles

ani = FuncAnimation(fig, update, frames=len(t), interval=50, blit=True)
plt.close()
HTML(ani.to_jshtml())
```
